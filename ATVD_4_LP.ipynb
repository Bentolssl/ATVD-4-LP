## 4) Atividade 2 — Qualidade dos dados

Agora investigue se existem problemas de qualidade.

Responda:

1. Existem valores nulos?
2. Em quais colunas eles aparecem?
3. Existem categorias escritas de formas diferentes para o mesmo canal de venda?

```python
display(df.isnull().sum())
print('\nSim, existem.')
print('\nAs colunas são: Canal_Venda, Preco_Unitario e Quantidade.')

display(df['Canal_Venda'].value_counts())
print('\nSim, existem.')
```

## 5) Atividade 3 — KPIs do negócio

Considere que você é analista de um varejista e precisa apresentar indicadores para a gestão.

Calcule os seguintes KPIs:

1. **Receita total**
2. **Lucro total**
3. **Ticket médio** (`Receita média por registro`)
4. **Quantidade total vendida**

Depois, escreva uma breve interpretação:
- O desempenho geral parece positivo?
- Receita e lucro caminham juntos?

```python
#receita total
print('Receita')
receita_total = df['Receita'].sum()
print('A receita total é de', receita_total)

#lucro total
print('\nLuco')
lucro_total = df['Lucro'].sum()
print('O lucro total é de', lucro_total)

#ticket médio
print('\nTicket')
ticket_medio = receita_total / df.shape[0]
print('O ticket médio é de', ticket_medio)

#Quantidade total vendida
print('\nQuantidade')
quantidade_total = df['Quantidade'].sum()
print('A quantidade total vendida é de', quantidade_total)

print('\nConclusão:')
print('O desempenho geral é positivo, porém, a receita e o \nlucro não seguem o mesmo sentido.')
```

## 6) Atividade 4 — Respondendo perguntas do gestor com dados

Use o dataset para responder:

1. **Qual canal gera mais receita?**
2. **Qual produto gera mais lucro?**
3. **Quais UFs concentram maior receita?**

Para cada resposta:
- mostre o código;

```python
print('Canal que gera mais receita:')
display(df.groupby('Canal_Venda')['Receita'].sum())


print('\nProduto que gera mais lucro:')
display(df.groupby('Produto')['Lucro'].sum())

print('\nEstados que geram maior receita:')
display(df.groupby('UF')['Receita'].sum())
```
